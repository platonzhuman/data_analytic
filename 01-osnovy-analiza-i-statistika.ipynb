# Анализ данных — занятие 1
### Основы анализа данных и статистика

**Логика занятия:** сначала — что делает аналитик, какие бывают данные и как их готовят (теория), затем — случайные величины, вероятность, распределения и их числовые характеристики (практика с кодом).

## 1. Кто такой аналитик данных

Аналитик данных — тот, кто извлекает смысл из беспорядочных данных.

Ключевые навыки:

- **Программирование** — знать инструменты, языки и библиотеки для обработки данных и получения аналитической информации.
- **Методы анализа** — знать, как очищать данные, обрабатывать их и извлекать смысл.
- **Визуализация** — уметь показать результат.

> Дополнительно по литературе: книжки 3, 4, 12, 13, 14 — что можете, то и читайте.

## 2. Виды анализа данных

| Вид | На какой вопрос отвечает |
| --- | --- |
| Описательный | «Что происходит?» |
| Диагностический | «Почему это произошло?» |
| Прогностический | «Что может произойти?» |
| Рекомендационный | «Что сделать, чтобы это произошло?» — самый сложный |

**Описательный** — самый простой: нужно визуализировать и описать данные.

- **Агрегирование** — собрать данные из множества источников в один.
- **Первичный анализ** — определить типы данных и пределы их изменений.

Шаги описательного анализа:

1. **Определить метрики** — ключевые параметры, которые хотим изучать.
2. **Найти данные**, необходимые для расчёта этих метрик.
3. **Извлечь и подготовить данные** — вытащить из разных источников, объединить и использовать дальше.

## 3. Типы данных

- **Неупорядоченные** множества.
- **Частично-упорядоченные** (с ними чаще всего работаем): множества векторов, множества многомерных изображений.
- **Упорядоченные**: одномерные или временные, многомерные или пространственно-временные, дробной размерности (фрактальные).

**По способу получения:**

1. Первичные (сырые)
2. Вторичные (обработанные)

## 4. Типы шкал

**Категориальные** (можно разбить на категории):

- качественно дискретные:
  - *номинальная* — можно посчитать, кого здесь больше;
  - *порядковая* — есть отношение порядка (a ≥ b).

**Числовые:**

- количественно непрерывные:
  - *интервальная* — заданы начало и конец, можно считать среднее («если тебе 18 лет, твой возраст от 18 до 19»);
  - *шкала отношений* — считаем соотношения, но только при точных входных данных.

## 5. Агрегация данных

Источники данных:

1. **Потоки** — область в памяти, на диск не сбрасывается. Обычно буфер 8 КБ; 0 — stdin, 1 — stdout, 2 — stderr. В Linux потоки бинарные, буфер создаётся командой `mkfifo`.
2. **Текстовые файлы** — работаем в основном с ними.
3. **Бинарные файлы.**
4. **Базы данных.**
5. **Протоколы.**
6. **Внешние устройства** — взаимодействие через драйверы.

Подходы к агрегации: матрицы, словари, XML/JSON, базы данных, файловые системы.

## 6. Приведение к одному масштабу

- **Прореживание** — уменьшает количество данных (это же и его недостаток).
- **Интерполяция** — плавное преобразование: одна часть сохраняется, другая дополняется (минус — нужно добавлять данные). Бывает линейная.
- **Аппроксимация** — убираем данные, но оставляем модель и потом её показываем. Бывает линейная.

## 7. Методы визуализации

- **Общая** — для числовых значений (диаграммы, схемы).
- **Специальная** — для конкретной информации (карты, полярные графики).
- **Концептуальная** — для сложных концепций.
- **Стратегическая** — переводит данные в визуальную форму.
- **Метафорическая** — информация в виде геометрических фигур.
- **Комбинированная** — всё вместе.

# Практика: случайные величины и распределения

## 8. Случайная величина и статистика

**Случайная величина** — величина, значение которой заранее предсказать нельзя.

Пример: бросаем кубик 6 раз и получаем `2 3 5 2 4 6`. Набор значений из эксперимента — это **выборка**; работаем всегда с конечной выборкой.

Если бросить монету один раз, вероятность исхода — 100% (один исход из одного).

**Статистика** — любая функция, вычисляемая по случайным величинам. Любая функция от случайной величины — тоже случайная величина.

## 9. Вероятность

**Вероятность** — количество удачных исходов / количество всех исходов (классическое определение). В машинном обучении вероятность — мера события.

Свойства (аксиомы Колмогорова):

1. `0 ≤ P ≤ 1`
2. Если события A и B не связаны: `P(A или B) = P(A) + P(B)`
3. Сумма вероятностей по всем возможным событиям равна 1.

Если величина категориальная — вероятность определить можно.

## 10. Функция распределения и плотность

- **Функция распределения (CDF)** `F(x₀)` — вероятность, что величина меньше `x₀`.
- Вероятность попасть в интервал `[x₀, x₀+d]` — это `F(x₀+d) − F(x₀)`.
- **Плотность вероятности (PDF)** `p(x)`.

## 11. Числовые характеристики

- **Среднее (математическое ожидание)**: `E[x] = (x₀ + x₁ + … + xₙ) / n`
- **Дисперсия**: `D = E[(x − E[x])²]`
- **СКО** — корень из дисперсии.
- **Медиана** — упорядочиваем значения и берём центральное; она не меняется от одного выброса, то есть **робастна** (устойчива).
- **Интерквартильный размах (IQR)** — делим данные на квартили и берём разницу между третьим и первым квартилем. *Квартá* по-латыни — четверть.

Посчитаем эти характеристики на выборке:

```python
%matplotlib inline
import numpy as np
import scipy.stats as ss
import matplotlib.pyplot as plt

np.random.seed(42)
plt.rcParams["figure.figsize"] = (8, 4)
```

```python
sample = ss.norm(loc=0, scale=1).rvs(size=1000)

print("среднее (мат. ожидание):", sample.mean())
print("дисперсия:              ", sample.var())
print("СКО:                    ", sample.std())
print("медиана:                ", np.median(sample))

q1, q3 = np.percentile(sample, [25, 75])
print("IQR (Q3 - Q1):          ", q3 - q1)
```

Медиана устойчива к выбросам, а среднее — нет:

```python
with_outlier = np.append(sample, 100)

print("среднее с выбросом: ", with_outlier.mean())
print("медиана с выбросом:", np.median(with_outlier))
```

## 12. Основные распределения

1. Бернулли
2. Биномиальное
3. Равномерное
4. Нормальное (два параметра)
5. Хи-квадрат

Построим каждое и посмотрим на форму.

### Распределение Бернулли — один опыт с двумя исходами

```python
bernoulli = ss.bernoulli(0.5)
x = bernoulli.rvs(size=1000)

plt.hist(x, bins=[-0.5, 0.5, 1.5], density=True, rwidth=0.8)
plt.title("Распределение Бернулли, p = 0.5")
plt.xlabel("значение")
plt.ylabel("доля")
plt.show()
```

### Биномиальное распределение — число успехов в n опытах

```python
binom = ss.binom(5, 0.7)
x = binom.rvs(size=1000)

plt.hist(x, bins=np.arange(-0.5, 6.5, 1), density=True, rwidth=0.8)
plt.title("Биномиальное распределение, n = 5, p = 0.7")
plt.xlabel("число успехов")
plt.ylabel("доля")
plt.show()
```

### Равномерное распределение

```python
uniform = ss.uniform()
x = uniform.rvs(size=10000)

plt.hist(x, bins=50, density=True)
plt.title("Равномерное распределение")
plt.xlabel("значение")
plt.ylabel("плотность")
plt.show()
```

Сумма квадратов двух равномерных величин уже не равномерна — распределение становится похожим на скошенное:

```python
x = uniform.rvs(size=(10000, 2))

plt.hist((x ** 2).sum(axis=-1), bins=100, density=True)
plt.title("Сумма квадратов двух равномерных величин")
plt.xlabel("значение")
plt.ylabel("плотность")
plt.show()
```

### Нормальное распределение — PDF и CDF

```python
normal = ss.norm(loc=0, scale=1)
x = normal.rvs(size=10000)
xs = np.linspace(-4, 4, 1000)

plt.hist(x, bins=60, density=True, histtype="step", label="выборка")
plt.plot(xs, normal.pdf(xs), label="PDF")
plt.plot(xs, normal.cdf(xs), label="CDF")
plt.title("Нормальное распределение, μ = 0, σ = 1")
plt.legend()
plt.show()
```

### Гамма-распределение (параметр формы `a`)

```python
a = 2.3
gamma = ss.gamma(a)
L = gamma.rvs(size=1000)

plt.hist(L, density=True, histtype="step", bins=100, label=f"Гамма, a = {a}")
plt.title("Гамма-распределение")
plt.legend()
plt.show()
```

### Распределение хи-квадрат

```python
chi2 = ss.chi2(df=3)
x = chi2.rvs(size=10000)

plt.hist(x, bins=60, density=True)
plt.title("Распределение хи-квадрат, df = 3")
plt.xlabel("значение")
plt.ylabel("плотность")
plt.show()
```
