```python
import numpy as np
import scipy.stats as ss
```

```python
# сами данные
d = np.loadtxt("files/input/ivaniv.dat.txt")
before = d[:, 0]
after = d[:, 1]
dif = after - before

print("среднее до =", round(before.mean(), 2))
print("среднее после =", round(after.mean(), 2))
print("ухудшили", (dif < 0).sum(), "| улучшили", (dif > 0).sum(), "| без изменений", (dif == 0).sum())
```

```python
# изменился ли средний рейтинг?
t, p = ss.ttest_rel(after, before)

print("t =", round(t, 3), " p =", round(p, 5))
print("вывод:", "изменился" if p < 0.05 else "не изменился", "вниз" if dif.mean() < 0 else "верх")
```

```python
# изменился уровень поддержки?
sb, sa = before >= 50, after >= 50

a  = int((sb & sa).sum())      
b  = int((sb & ~sa).sum())     
c  = int((~sb & sa).sum())     
dd = int((~sb & ~sa).sum())   

n = a + b + c + dd
Z = (c - b) / np.sqrt((b + c) - (b - c) ** 2 / n)
p_s = 2 * (1 - ss.norm.cdf(abs(Z)))

print(f"поддержка до {100*sb.mean():.1f}%  после {100*sa.mean():.1f}%")
print(f"таблица: a={a} b={b} c={c} d={dd}")
print("Z =", round(Z, 3), " p_s =", round(p_s, 4))
print("вывод:", "изменилась" if p_s < 0.05 else "не изменилась")
```
